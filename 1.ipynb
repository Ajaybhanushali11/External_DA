```python
# #import library and load data sets 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

movies= pd.read_csv('Movie_CSV1.csv')
shows= pd.read_csv('Movie_CSV2.csv')
bookings= pd.read_csv('Movie_CSV3.csv')

print("all datasets Loaded Succesfully")

#dispaly fast
print(movies.head(5))
print(shows.head(5))
print(bookings.head(5))

#Display Last
print(movies.tail(5))
print(shows.tail(5))
print(bookings.tail(5))

#dataset Dimension ,Structure,and data type

#Dimension
print("Movie shape",movies.shape)
print("Shows Shape",shows.shape)
print("Booking Shape", bookings.shape) 

#structure
movies.info()
shows.info()
bookings.info()

#Data Type

print(movies.dtypes)
print(shows.dtypes)
print(bookings.dtypes)

#s.4 Descriptive Statistics
print(movies.describe(include='all'))
print(shows.describe(include='all'))
print(bookings.describe(include='all'))

#s.5 missing value
print(movies.isnull().sum())
print(shows.isnull().sum())
print(bookings.isnull().sum())

#s.6 missing value Handling
print(movies.isnull().sum())
print(movies.isnull().sum())

# Handle Missing Values
movies['Genre'] = movies['Genre'].fillna('Unknown')
movies['Language'] = movies['Language'].fillna('Unknown')
movies['Movie_Status'] = movies['Movie_Status'].fillna('Unknown')
movies['Director'] = movies['Director'].fillna('Unknown')

movies['Rating'] = movies['Rating'].fillna(movies['Rating'].median())

movies['Duration_Min'] = movies['Duration_Min'].fillna(
    movies['Duration_Min'].median()
)

movies['Production_Budget'] = movies['Production_Budget'].fillna(
    movies['Production_Budget'].median()
)

print("Movies Missing Values After Cleaning:")
print(movies.isnull().sum())

#Show Table 
# Check Missing Values
print(shows.isnull().sum())

# Handle Missing Values
shows['Cinema_Name'] = shows['Cinema_Name'].fillna('Unknown')
shows['City'] = shows['City'].fillna('Unknown')

shows['Total_Seats'] = shows['Total_Seats'].fillna(
    shows['Total_Seats'].median()
)

shows['Ticket_Price'] = shows['Ticket_Price'].fillna(
    shows['Ticket_Price'].median()
)

print("Shows Missing Values After Cleaning:")
print(shows.isnull().sum())

#Bookings table
# Check Missing Values
print(bookings.isnull().sum())

# Handle Missing Values
bookings['Customer_Gender'] = bookings['Customer_Gender'].fillna('Unknown')
bookings['Payment_Mode'] = bookings['Payment_Mode'].fillna('Unknown')
bookings['Booking_Channel'] = bookings['Booking_Channel'].fillna('Unknown')

bookings['Customer_Age'] = bookings['Customer_Age'].fillna(
    bookings['Customer_Age'].median()
)

bookings['Tickets'] = bookings['Tickets'].fillna(1)
bookings['Discount'] = bookings['Discount'].fillna(0)
bookings['Convenience_Fee'] = bookings['Convenience_Fee'].fillna(30)

print("Bookings Missing Values After Cleaning:")
print(bookings.isnull().sum())

#s.7 Duplicate remove 
print(movies.duplicated().sum())
print(shows.duplicated().sum())
print(bookings.duplicated().sum())
# #remove
movies=movies.drop_duplicates()
shows=shows.drop_duplicates()
bookings =bookings.drop_duplicates()

#Duplicarion id check
print(movies['Movie_ID'].duplicated().sum())
print(shows['Show_ID'].duplicated().sum())
print(bookings.head(5))
# print(bookings['Bookings_ID'].duplicated().sum())
```

```python
#s.11 Text  Cleaning and Standardization
# Movies
movies['Genre'] = movies['Genre'].str.strip().str.title()
movies['Language'] = movies['Language'].str.strip().str.title()
movies['Director'] = movies['Director'].str.strip().str.title()

# Shows
shows['Cinema_Name'] = shows['Cinema_Name'].str.strip().str.title()
shows['City'] = shows['City'].str.strip().str.title()
shows['Show_Type'] = shows['Show_Type'].str.strip().str.upper()

# Bookings
bookings['Customer_Gender'] = bookings['Customer_Gender'].str.strip().str.title()
bookings['Payment_Mode'] = bookings['Payment_Mode'].str.strip().str.title()
bookings['Booking_Status'] = bookings['Booking_Status'].str.strip().str.title()
bookings['Booking_Channel'] = bookings['Booking_Channel'].str.strip().str.title()
```

```python
# invalid  and abnormao value identifid

#Movie rating should be 0 and 10

movies.loc[movies['Rating']>10,'Rating'] = np.nan
movies['Rating'] = movies['Rating'].fillna(movies['Rating'].median())

#movie Duration
print(movies[movies['Duration_Min']<=0])

#Customer Age Should be positive
print(bookings[bookings['Customer_Age']<=0])
bookings.loc[bookings['Customer_Age']<0,'Customer_Age'] = np.nan

# Ticket Price should be positive
print(shows[shows['Price']<=0])
shows.loc[shows['Price']<=0,'Price']= np.nan

#s.13 date Conversion and Year extraction

movies['Release_Date'] = pd.to_datetime(movies['Release_Date'],errors='coerce')
shows['Show_Date'] = pd.to_datetime(shows['Show_Date'],errors='coerce')
bookings['Booking_Date'] = pd.to_datetime(bookings['Booking_Date'],errors='coerce')

#extract year
movies['Release_Year'] = movies['Release_Date'].dt.year
shows['show_Year']=shows['Show_Date'].dt.year
bookings['Booking_Date'] = bookings['Booking_Date'].dt.year

print(movies[['Release_Date','Release_Year']].head())

#s 15 Mathematical Cal
#Total boking amount
bookings['Calculated_Amount']=bookings['Ticket_Amount']
- bookings['Discount']
+ bookings['Convenience_Fee']

#discount per

bookings['Discount_Percentage'] = (
    bookings['Discount']/bookings['Ticket_Amount']*100

)
print(bookings[['Ticket_Amount', 'Discount', 'Calculated_Amount']].head())

#s 16 group and aggregation

#movies by genre
print(movies.groupby('Genre')['Rating'].mean())

#Average ticket price by City
print(shows.groupby('City')['Price'].mean())

#Total bookings amount by payment mode
print(bookings.groupby('Payment_Mode')['Final_Amount'].sum())

#booking Sattus count
print(bookings.groupby('Booking_Status')['Booking_ID'].count())
```

```python
#s 17 Data set Merging
#Merge movie and Show using Movie_ID
movie_show=pd.merge(
    movies,shows,on='Movie_ID',
    how='inner'
)
#print(movie_show.head())
#Merge SHow and Bookings using Show_ID

show_bookings = pd.merge(
    shows,bookings,on='Show_ID',
    how="inner"
)
print(show_bookings.head())
#Merge al three Datasets
final_data =pd.merge(
    movie_show,
    bookings,
    on='Show_ID',
    how='inner'
)
print(final_data.head())
#s 18 Splitting and Recombining dataset

#split movie int ot two part 
part1 =movies.iloc[:250]
part2 = movies.iloc[250:]

print(part1.shape)
print(part2.shape)

#Recombine 
combine_movies = pd.concat([part1,part2],ignore_index=True)
print(combine_movies.shape)
```

```python
movies.to_csv('Cleaned_Movies.csv', index=False)

shows.to_csv('Cleaned_Shows.csv', index=False)

bookings.to_csv('Cleaned_Bookings.csv', index=False)

final_data.to_csv('Final_Movie_Booking_Data.csv', index=False)

print("All datasets exported successfully")
```
